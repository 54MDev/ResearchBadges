```python
from sklearn.datasets import load_breast_cancer
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import confusion_matrix, make_scorer
from sklearn.model_selection import cross_validate

# Load dataset and initialize your model with your exact settings
data = load_breast_cancer()
X, y = data.data, data.target

clf = RandomForestClassifier(
    n_estimators=200, max_depth=10, max_features='sqrt', random_state=42
)


# Define custom specificity scorer
def specificity_score(y_true, y_pred):
  tn, fp, fn, tp = confusion_matrix(y_true, y_pred).ravel()
  return tn / (tn + fp)


# Map multiple metrics as demonstrated in section 3.1.1.1
scoring = {
    'auc': 'roc_auc',
    'sensitivity': 'recall',
    'specificity': make_scorer(specificity_score),
}

scores = cross_validate(clf, X, y, cv=5, scoring=scoring)

print('AUC:', scores['test_auc'].mean())
print('Sensitivity:', scores['test_sensitivity'].mean())
print('Specificity:', scores['test_specificity'].mean())
```

```python
#LOSO-CV
## Synthetic group ID array was made just for completion's sake.
## The breast cancer dataset doesn't have group IDs. It's a binary dataset like my OCD one though.

import numpy as np
from sklearn.model_selection import LeaveOneGroupOut

# Create mock group IDs (e.g., dividing the 569 samples across ~30 patients/groups)
groups = np.random.randint(0, 30, size=X.shape[0])

logo = LeaveOneGroupOut()

# Now this will run without a NameError
scores = cross_validate(
    clf, X, y, cv=logo.split(X, y, groups=groups), scoring=scoring
)

print('Mean AUC:', scores['test_auc'].mean())
print('Mean Sensitivity:', scores['test_sensitivity'].mean())
print('Mean Specificity:', scores['test_specificity'].mean())
```
